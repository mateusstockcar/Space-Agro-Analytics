```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Gerando dados orbitais simulados (Mock Data para a POC)
# Matrizes que representam a refletância de luz de uma plantação
tamanho = (500, 500)
nir_band = np.random.uniform(0.5, 0.9, tamanho) # Banda Infravermelha (Vegetação saudável reflete muito)
red_band = np.random.uniform(0.1, 0.4, tamanho) # Banda Vermelha (Vegetação saudável absorve)

# Simulando uma anomalia: área com estresse hídrico (seca) no centro da fazenda
nir_band[200:300, 200:300] = 0.3 # Cai o reflexo do infravermelho
red_band[200:300, 200:300] = 0.6 # Aumenta o reflexo do vermelho

# 2. Aplicando o Algoritmo de NDVI Computacional
# Fórmula padrão da agricultura de precisão
ndvi = (nir_band - red_band) / (nir_band + red_band)

# 3. Visualização B2B Premium
plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=(10, 8))

# O colormap 'RdYlGn' (Red-Yellow-Green) é o padrão da indústria agro para NDVI
cax = ax.imshow(ndvi, cmap='RdYlGn', vmin=-1, vmax=1)

# Estilização B2B (Dark Mode com acentos premium)
ax.set_title("Space Agro Analytics - Análise Orbital (NDVI)", fontsize=18, color='#FFD700', pad=20) # Dourado
ax.axis('off') # Remove os eixos para parecer um dashboard

# Barra de legenda customizada
cbar = fig.colorbar(cax, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label('Índice de Vegetação Saudável', color='#00BFFF', fontsize=12) # Azul elétrico
cbar.ax.tick_params(colors='white')

plt.tight_layout()
plt.show()
```
